```python
df["airline"].value_counts(ascending=True).plot.barh(color=["lightpink","lightblue"])
plt.title("AIRLINES WITH FREQUENCIES")
plt.xlabel("number of counts")
plt.ylabel("AIRLINES NAMES")
plt.show()
```
